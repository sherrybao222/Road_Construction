```python
home_dir = '/Users/dbao/Library/CloudStorage/GoogleDrive-sherrydbao@gmail.com/.shortcut-targets-by-id/1sYZNVBbYCrHcHjo-9xdKcA4wLnob9ZEq/'+'/road_construction/data/2022_online/'
data_dir = home_dir + 'data/demographics/'
data_dir2 = home_dir + 'R_analysis_data/'

df = pd.read_csv(data_dir + 'demographics_subjects.csv')
data = pd.read_csv(data_dir2 + 'data.csv')

demo_filtered = df[df["Prolific ID"].isin(data["ID_original"])]

print("Original demographics:", len(df))
print("Filtered demographics:", len(demo_filtered))
```

```python
# Clean and prepare
demo_filtered['age'] = pd.to_numeric(demo_filtered['age'], errors='coerce')  # convert age to numeric
demo_filtered['Gender'] = demo_filtered['Gender'].str.strip().str.title()    # standardize capitalization
demo_filtered['Race'] = demo_filtered['How would you describe your race or ethnicity? Select one or more: - Selected Choice'].str.strip()

# Summaries
print("=== Gender Distribution ===")
print(demo_filtered['Gender'].value_counts(dropna=False))
print()

print("=== Age Summary ===")
print(demo_filtered['age'].describe())  # count, mean, std, min, 25%, 50%, 75%, max
print()

# Optional: group ages into bins
bins = [0, 18, 25, 35, 50, 65, 100]
labels = ["<18", "18-24", "25-34", "35-49", "50-64", "65+"]
demo_filtered['AgeGroup'] = pd.cut(demo_filtered['age'], bins=bins, labels=labels, right=False)
print(demo_filtered['AgeGroup'].value_counts().sort_index())
print()

print("=== Race / Ethnicity Distribution ===")
print(demo_filtered['Race'].value_counts(dropna=False))
```
