## supp: Remaining budget

```python
# Create 5 bins for each subject individually
subject_bins = []
new_df = pd.DataFrame()
for subject in undo_data['subjects'].unique():
    subject_data = undo_data[undo_data['subjects'] == subject].copy()
    
    # Create 5 equal-width bins for this subject's log RTs
    temp1, bin_edges = pd.cut(subject_data['leftover'], bins=10, retbins=True, labels=False)
    subject_data['leftover_bin'] = temp1
    new_df = pd.concat([new_df, subject_data])

    bin_means = subject_data.groupby('leftover_bin')['leftover'].mean()
    for bin_num in range(10):
        if bin_num in bin_means.index:  # Check if bin exists for this subject
            subject_bins.append({
                'subjects': subject,
                'bin': bin_num,
                'mean_leftover': bin_means[bin_num]
            })

undo_data = new_df

# Convert to DataFrame
binned_results = pd.DataFrame(subject_bins)
final_results = binned_results.groupby('bin').agg({
    'mean_leftover': ['mean', 'sem']
}).reset_index()
print(final_results)

# replace bin with mean_log_rt
# Create a mapping dictionary from bin number to mean_log_rt
bin_to_mean_leftover_mapping = final_results.set_index('bin')['mean_leftover'].to_dict()['mean']
# Apply the mapping, with a fallback for any NaN values
undo_data['leftover_bin_mean_leftover'] = undo_data['leftover_bin'].map(bin_to_mean_leftover_mapping)
```

# Figure 4c: undoRT

```python
def organize_data(df, cond, type_):
    index_singleUndo = df.index[cond]
    df1 = df.loc[index_singleUndo, ['subjects', 'puzzleID', 'allMAS','RT', 'undo', ]]
    df1["N_more"] = df.loc[index_singleUndo-1, 'N_more'].values
    df1["currNumCities"] = df.loc[index_singleUndo-1, 'currNumCities'].values
    df1["severityOfErrors"] = df.loc[index_singleUndo-1, 'severityOfErrors'].values
    df1["cumulative_error"] = df.loc[index_singleUndo-1, 'cumulative_error'].values
    df1["scaled_N_more_bin"] = df.loc[index_singleUndo-1, 'scaled_N_more_bin'].values
    df1["checkEnd"] = df.loc[index_singleUndo-1, 'checkEnd'].values
    df1['type'] = type_
    return df1

df1 = organize_data(sc_data_choice_level, (sc_data_choice_level['firstUndo'] == 1) &(sc_data_choice_level['lastUndo'] == 1), "singleUndo")
df2 = organize_data(sc_data_choice_level, (sc_data_choice_level['firstUndo'] == 1) &(sc_data_choice_level['lastUndo'] != 1), "seqUndo")
df3 = organize_data(sc_data_choice_level, (sc_data_choice_level['undo'] == 0) & (sc_data_choice_level['RT'] != -1) & (sc_data_choice_level['submit'] == 0), "move")

df_all = pd.concat([df1, df2, df3])
```

```python
%%R -i df_all,out_dir2
df_all$subjects <- as.factor(df_all[,"subjects"])
df_all$puzzleID <- as.factor(df_all[,"puzzleID"])
df_all$undo <- as.factor(df_all[,"undo"])
df_all$RT <- df_all$RT/1000
df_all$RT_log <- log(df_all$RT)

write.csv(df_all, file = paste0(data_dir, "statsdata_rt.csv"), row.names = F)
```
